###Find Greatest among three numbers

```python
a=int(input("Enter a value"))
b=int(input("Enter another value"))
c=int(input("Enter another value"))
if (a>b and a>c):
  print(f"{a} is greater")
elif(b>a and b>c):
  print(f"{b} is greater")
else:
  print(f"{c} is greater")
```

###print FIZZ if a number is divisible by 3, print BUZZ if a number is divisible by 5 and print FIZZBUZZ if the number is divisible by both 3 and 5

```python
a=int(input("Enter a number"))
if(a%3==0 and a%5==0):
  print("FIZZBUZZ")
elif(a%5==0):
  print("BUZZ")
elif(a%3==0):
  print("FIZZ")
```

###program that swaps two variables without using temp variables

```python
a=int(input("Enter a value"))
b=int(input("Enter b value"))
a,b=b,a
print(f"a is {a} and b is {b}")
```

###Switch case use case

```python
a=int(input("Enter a number < 11 --> "))
match a:
  case 1|3|5|7|9 :
    print(f"{a} is odd")
  case 2|4|6|8|10:
    print(f"{a} is even")
  case _:
    print("out of bound")
```
