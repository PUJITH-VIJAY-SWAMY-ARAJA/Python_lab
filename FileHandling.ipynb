```python
#we can also open a file using with statement as follows:
#syntax: with open("file_name","mode") as our filename project(f1):

with open("sample.txt","r") as f1:
  content=f1.read()
  #f1.read() is used to read all the characters in a file f1
  print(content)
```

```python
with open("sample.txt","r") as f1:
  content1=f1.read(25) # used to read 8 characters
  print(content1)
```

```python
with open("sample.txt","r") as f1:
  content2=f1.readlines() # used to read all the lines in a file in a list format
  print(content2)
```

```python
with open("sample.txt","r") as f1:
  content3=f1.read(15)
  print(f1.tell())
  print(content3)
  #f1.tell() is a method used to tell the position of the pointer or cursor
```

```python
with open("sample.txt","r") as f1:
  content4=f1.read(15)
  print(f1.tell())
  print(content4)
  f1.seek(0) #used to change the position of the cursor or the pointer
  print(f1.tell())
  print(f1.read())
```

```python
#if we only want to print a single line then we use readline() method
with open("sample.txt","r") as f1:
  print(f1.readline())
  print(f1.readlines())
#readlines() used to read everyline in a list
```

Write a program to sort words in a file and put them in another file. The output file should have only lowercase words so any uppercase words from source must be lowered.

```python
with open("file2.txt","w") as f:
  f.write("Python JAVA apple Banana Red")
  print("file created and content added successfully")
with open("file2.txt","r") as  f1:
  data=f1.read().split() #used to collect in a list
  data_lower=[i.lower() for i in data]
  sort_data=sorted(data_lower)
with open("new2.txt","w") as f2:
  for i in sort_data:
    f2.write(i+"\n")
```

Write a python program that counts number of lines,number of words,and number of characters in a text file

```python
with open("file2.txt","r") as f:
  lines=f.readlines()
  print(len(lines))
  words=0
  chars=0
  for i in lines:
    words=words+len(i.split())
    chars=chars+len(i)
  print(words)
  print(chars)
```

 Binary Files Handling

 Serialization: The process of converting binary data into a byte data is called as serialization.

 Builtin Module we use pickle module

 In pickle module we use two methods:
 1. Dumping 
 2. Loading

 pickle.dump():creates a file - used to write the object data into the file

 pickle.load():reads a file-used to read the object data and get into the code.

```python
import pickle
L=[1,2,3]
f=open("file3.txt","wb")
pickle.dump(L,f)
f.close()
#Garbage values in the file which is only understand by machine
```

```python
import pickle
f=open("file3.txt","rb")
pickle.load(f)
f.close()
```

Handling CSV Files

In python we have a module called csv(comma seperated values) used to handle table kind of data.

```python
#creating a csv file
import csv
f=open("table.csv","w")
w=csv.writer(f)
w.writerow(["s.no","name","age"])
w.writerows([["1","ajay","18"],["2","veera","21"]])
f.close()
```

```python
#Reading a csv file
f=open("table.csv","r")
r=csv.reader(f) #Row wise list data
for i in r:
  print(i)
f=open("table.csv","r")
r=csv.DictReader(f) #DictReader()-Data format in dictionaries
for i in r:
  print(i)
```
