#**STRINGS**

* Set of characters is called a string.
* Strings are immutable in nature



```python
#Most useful Method
a="hello world from python"
b=a.split()
print(b)
```

```python
c=" ".join(b)
print(c)
c="".join(b)
print(c)
```

```python
#Strip() is used to remove trailing spaces(front and back) in the string.
x=" AI & DS "
print(x.strip())
```

##String Methods
---
1. split()
2. join()
3. upper() and lower()
4. strip()
5. replace()
6. find()
7. count()

```python
text="AI is easy"
print(text.split())
print(text.split(","))
print(text.upper())
print(text.lower())
print(text.replace("easy","hard"))
print(text.find("hard"))
print(text.count("is"))
print(text)
```
