#**Tuples**

###creating a tuple and operations on tuple




```python
t=(1,2,3) #this is the creation of tuple
print(t)
```

##userdefined tuple

```python
tu=tuple(int(i) for i in input().split())
print(tu)
```

##operations on tuple

```python
t1=(1,2,4,6,3,7,3)
print(t+t1) #concatenation
print(t*3) #repeatition
if 1 in t1:
  print("YES")
else:
  print("NO")
print(t1[::-1])#reverse of a tuple
```

###Functions in Tuple

```python
t1=tuple(int(i) for i in input().split())
print(max(t1))
print(min(t1))
print(sum(t1))
print(tuple(sorted(t1)))
```

###Methods in Tuple
COUNT,INDEX

```python
print(t1.count(1)) ##it counts a number, how many times repeated
print(t1.index(3))
```

####ZIP function

```python
t1=(1,2,3,8) # here we can see irresepective of the size the zip will only zip the same size.
t2=(1,3,5)
print(tuple(zip(t1,t2)))
```

###operations on sets

```python
s1={1,2,3}
s2={1,2,3}
print(s1|s2)
print(s1&s2)
print(s1-s2)
print(s1^s2)
```

```python
#To change the original set
c={11,22,33}
a={1,2,3,33}
print(a.union(c))
print(a.difference(c))
print(a.intersection(c))
print(a.intersection_update(c))
print(a) # a is intersection set contains only 33
```

PROGRAM 2 : Write a python program to find  Number of duplicates in the list

```python
L=list(map(int,input().split()))
K=set(L)
print(len(L)-len(K))
```

Finding Duplicate elements

```python
L1=[i for i in input().split()]
L2=[]
for i in L1:
  if L1.count(i)>1:
    L2.append(i)
L3=set(L2)
print(len(L3))
```
