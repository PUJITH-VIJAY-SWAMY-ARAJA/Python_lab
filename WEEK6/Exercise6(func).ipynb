###Program that checks whether the given number is palindrome or not using functions.

```python
def palindrome(n):
  l=0
  s=n
  while s!=0:
    l = l*10 + (s%10)
    s//=10
  if l==n:
    return True
  else:
    return False
a=int(input())
res=palindrome(a)
if res:
  print("YES")
else:
  print("NO")
```

###program that reverse a given string using functions

```python
def revStr(n):
  rev=""
  for i in range(len(n)-1,-1,-1):
    rev=rev+n[i]
  print(rev)
a=str(input())
revStr(a)
```

####Example on variable length aruguments

```python
def add(*num):
  print(sum(num))
add(1,2,3,4,5,6,7)
```

####Example on variable length of arguments via throughput and taking multiple values as a output

```python
def Mvalues(*num):
  add=sum(num)
  maxi=max(num)
  mini=min(num)
  return add,maxi,mini
madd,mmaxi,mmini=Mvalues(1,2,3,4,56)
print(f"add is {madd} maximum is {mmaxi} and minimum is {mmini}")
```
