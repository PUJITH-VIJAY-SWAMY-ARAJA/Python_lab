##print 1 to n odd numbers

```python
n=int(input())
for i in range(1,n,2):
  print(i,end=" ")
```

##print multiplication table for a given number

```python
n=int(input())
for i in range(1,11,1):
  print(n,"*",i,"=",n*i)
```

##check whether given number is a prime or not

```python
n=int(input())
count=0
for i in range(1,n):
  if n%i == 0:
    count+=1
if(count!=1):
  print("not prime")
else:
  print("prime")


#for i in range(2,n):
#  if n%i ==0:
#    print("it is not prime")
#    break
#else:
#  print("prime")
```

##print prime numbers between 2 numbers

```python
n=int(input("Enter starting number: "))
l=int(input("Enter last number: "))
for i in range(n,l+1):
  for j in range(2,i):
    if i%j ==0:
      break
  else:
    print(i,end=" ")
```

##print sum of the digits in a number

```python
n=int(input("Enter a number: "))
sum = 0
while(n!=0):
  sum=sum+(n%10)
  n=n/10
print("sum is",int(sum))
```
